# Cacao Flavor Rating Prediction
This notebook follows the same ML workflow as the supplied Churn Prediction project, adapted to the Flavors of Cacao dataset. Target: `Rating`.

```
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('../data/raw/flavors_of_cacao.csv')
df.head()
```

```
print(df.shape)
print(df.info())
df.describe(include='all').T
```

```
df['Rating'].value_counts().sort_index().plot(kind='bar', title='Rating Distribution')
plt.xlabel('Rating'); plt.ylabel('Count'); plt.show()
```
